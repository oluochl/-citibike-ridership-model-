## C1: Fix the Dtypes

```python

import pandas as pd
import numpy as np


data = pd.read_csv('/Users/laven/Documents/TKH/-citibike-ridership-model-/Data/citibike_weather_daily.csv', 
    parse_dates=["ride_date","obs_date"])               

numeric_columns = data[['num_rides','day', 'month', 'temp_f', 'max_temp_f', 'min_temp_f', 
'wind_speed_knots', 'precip_in' ]] 

for column in numeric_columns:
    if column in data.columns:
        data[column] = pd.to_numeric(data[column])

print(data.dtypes)
```

## C2: Handle the Coded Missing Values



```python

data['precip_in'] = data['precip_in'].replace(99.99, np.nan)

missing_precip = data['precip_in'].isna()

print(f"Total missing precipitation rows: {missing_precip.sum()}")

rows_with_missing_precip = data[missing_precip]
print(rows_with_missing_precip)

data = data.dropna(subset=['precip_in'])

```

Only one row is affected in the entire dataset. I have used dropna() to remove the affected row of data.

## C3: Encode Day of Week

```python
pd.get_dummies(data['day_of_week'], drop_first= False )
```

drop_first() removes the first day of the week of the days_of_week column.

```python
data.to_csv('/Users/laven/Documents/TKH/-citibike-ridership-model-/Data/citibike_weather_daily_clean.csv', index=False)
```
